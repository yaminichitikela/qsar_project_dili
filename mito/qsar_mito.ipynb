## Step 2 — Feature Engineering

### 2A. Morgan Fingerprints (ECFP4) — for Classical ML
### 2B. Rich Atom/Bond Graph Features — for GNN


````python
# ── 5. Morgan ECFP4 (2048 bits) ─────────────────────────────────────────────
def smiles_to_ecfp4(smi, radius=2, n_bits=2048):
    mol = Chem.MolFromSmiles(smi)
    if mol is None: return None
    fp  = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=n_bits)
    arr = np.zeros(n_bits, dtype=np.float32)
    DataStructs.ConvertToNumpyArray(fp, arr)
    return arr

fps = df['csmi'].apply(smiles_to_ecfp4)
valid = fps.notna()
df    = df[valid].reset_index(drop=True)
fps   = fps[valid]
X_fp  = np.vstack(fps.values)   # (N, 2048)
y_all = df['label'].values
print(f'Fingerprint matrix : {X_fp.shape}')
print(f'FP density         : {X_fp.mean():.4f}')
````

````python
# ──  Rich Atom Features (14-dim) & Bond Features (6-dim) — for GNN ─────────
#
#  14 atom features 
#    Atomic num one-hot (C,N,O,S,F,P,Cl,Br,I,other=10),
#    degree (0-5), formal charge, hybridization one-hot (SP/SP2/SP3/other=4),
#    is_aromatic, is_in_ring, total_Hs (0-4), normalized mass
#
#  6 bond features:
#    bond type one-hot (single/double/triple/aromatic=4),
#    is_conjugated, is_in_ring
#
ATOM_SYMBOLS = [6,7,8,16,9,15,17,35,53]  # C N O S F P Cl Br I

def one_hot(val, choices):
    vec = [0]*len(choices)
    try: vec[choices.index(val)] = 1
    except ValueError: pass   # 'other' → all zeros
    return vec

from rdkit.Chem import rdchem
HYBRID_MAP = {
    rdchem.HybridizationType.SP  : 0,
    rdchem.HybridizationType.SP2 : 1,
    rdchem.HybridizationType.SP3 : 2,
}

def atom_feat(atom):
    at_oh  = one_hot(atom.GetAtomicNum(), ATOM_SYMBOLS) + [0]  # 10-dim (+other)
    hyb    = [0,0,0,0]
    hyb[HYBRID_MAP.get(atom.GetHybridization(), 3)] = 1
    deg    = min(atom.GetDegree(), 5) #no of bonds this atom has
    deg_oh = [0]*6; deg_oh[deg]=1
    fc     = atom.GetFormalCharge() #chareg positve or nehative or zero
    fc_enc = [1 if fc<0 else 0, 1 if fc==0 else 0, 1 if fc>0 else 0]
    extras = [
        int(atom.GetIsAromatic()),
        int(atom.IsInRing()),
        min(atom.GetTotalNumHs(), 4) / 4.0,
        atom.GetMass() / 100.0
    ]
    return at_oh + deg_oh + fc_enc + hyb + extras  # 10+6+3+4+4 = 27-dim

BOND_TYPES = [
    rdchem.BondType.SINGLE,
    rdchem.BondType.DOUBLE,
    rdchem.BondType.TRIPLE,
    rdchem.BondType.AROMATIC
]

def bond_feat(bond):
    bt = one_hot(bond.GetBondType(), BOND_TYPES)  # 4-dim
    return bt + [int(bond.GetIsConjugated()), int(bond.IsInRing())]  # 6-dim

# Test dimensions
from rdkit.Chem import MolFromSmiles
_mol = MolFromSmiles('CC(=O)O')
ATOM_DIM = len(atom_feat(_mol.GetAtomWithIdx(0)))
BOND_DIM  = len(bond_feat(_mol.GetBondWithIdx(0)))
print(f'Atom feature dim : {ATOM_DIM}')
print(f'Bond feature dim : {BOND_DIM}')
````

````python
# ── 7. Build Graph Dataset ───────────────────────────────────────────────────
def mol_to_graph(smi, label):
    mol = Chem.MolFromSmiles(smi)
    if mol is None: return None
    # Atom features
    x = torch.tensor([atom_feat(a) for a in mol.GetAtoms()], dtype=torch.float)
    # Bond features + bidirectional edges
    ei, ea = [], []
    for bond in mol.GetBonds():
        i, j = bond.GetBeginAtomIdx(), bond.GetEndAtomIdx()
        bf = bond_feat(bond)
        ei += [[i,j],[j,i]]
        ea += [bf, bf]
    if ei:
        edge_index = torch.tensor(ei, dtype=torch.long).t().contiguous()
        edge_attr  = torch.tensor(ea, dtype=torch.float)
    else:
        edge_index = torch.zeros((2,0), dtype=torch.long)
        edge_attr  = torch.zeros((0, BOND_DIM), dtype=torch.float)
    y = torch.tensor([label], dtype=torch.float)
    return Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y)

graph_list = []
for smi, lbl in zip(df['csmi'].values, y_all):
    g = mol_to_graph(smi, lbl)
    if g: graph_list.append(g)

print(f'Graphs built: {len(graph_list):,}')
print(f'Example: {graph_list[0]}')
````

## Step 4 — Evaluation Helpers


````python
# ── 12. Metrics Helper ────────────────────────────────────────────────────────
def evaluate(y_true, y_prob, name, cpt=0.5):
    y_pred = (y_prob >= cpt).astype(int)
    tn,fp,fn,tp = confusion_matrix(y_true, y_pred).ravel()
    sens = tp/(tp+fn) if (tp+fn)>0 else 0
    spec = tn/(tn+fp) if (tn+fp)>0 else 0
    return {
        'Model'      : name,
        'CPT'        : cpt,
        'Sensitivity': round(sens*100,1),
        'Specificity': round(spec*100,1),
        'CCR'        : round((sens+spec)/2*100,1),
        'ROC-AUC'    : round(roc_auc_score(y_true, y_prob),4),
        'PR-AUC'     : round(average_precision_score(y_true, y_prob),4),
        'MCC'        : round(matthews_corrcoef(y_true, y_pred),4),
        'Bal-Acc'    : round(balanced_accuracy_score(y_true,y_pred)*100,1),
        'F1'         : round(f1_score(y_true,y_pred),4),
        'TP':tp,'TN':tn,'FP':fp,'FN':fn
    }

METRIC_COLS = ['Model','CPT','Sensitivity','Specificity','CCR',
               'ROC-AUC','PR-AUC','MCC','Bal-Acc','F1']
print('Helpers ready.')
````

## Step 5 — Classical ML: Hyperparameter Tuning on Validation Set

## Step 6 — Graph Neural Networks (Classification)

**Architecture** (same for all 4 variants):
```
SMILES  Atoms+Bonds (27-dim / 6-dim)
  - Linear projection (hidden)
  - 3 × [Conv + BatchNorm1d + ReLU + Dropout + Residual]
  - GlobalMeanPool
  - MLP head [hidden→hidden/2→1]
  - BCEWithLogitsLoss
```
Tuned hyperparameters per architecture: `hidden`, `n_layers`, `dropout`, `lr`  
Selection criterion: **val ROC-AUC** | Early stopping patience=20 epochs


````python
# ── 17. GNN Architecture ─────────────────────────────────────────────────────
class MolGNN(nn.Module):
    """
     GNN for molecular binary classification.
     GCN, GAT, GIN, GraphSAGE via conv_type argument.
    3 conv layers with BatchNorm + Residual connections.
    """

    def __init__(self, in_dim, hidden=128, n_layers=3, dropout=0.3,
                 conv_type='GCN', heads=4):
        super().__init__()
        self.conv_type = conv_type
        self.n_layers  = n_layers
        self.drop      = nn.Dropout(dropout)

        # Input projection
        self.proj = nn.Linear(in_dim, hidden) #linear layer to project input features to hidden dim
        self.bn0  = nn.BatchNorm1d(hidden) #batch norm after input projection

        # Conv layers + BatchNorm
        self.convs = nn.ModuleList() #list to hold convolutional layers
        self.bns   = nn.ModuleList() #list to hold batch norm layers after each conv
        for _ in range(n_layers): 
#             GCNConv(128, 128):
# Each atom collects messages from neighbours
# → takes weighted average
# → updates its own 128 numbers

            if conv_type == 'GCN':
                self.convs.append(GCNConv(hidden, hidden))
            elif conv_type == 'GAT':
                # concat=False → output stays at hidden dim
                self.convs.append(GATConv(hidden, hidden, heads=heads,
                                          concat=False, dropout=dropout))
            elif conv_type == 'GIN':
                mlp = nn.Sequential(
                    nn.Linear(hidden, hidden*2),
                    nn.BatchNorm1d(hidden*2),
                    nn.ReLU(),
                    nn.Linear(hidden*2, hidden))
                self.convs.append(GINConv(mlp, train_eps=True))
            elif conv_type == 'SAGE':
                self.convs.append(SAGEConv(hidden, hidden))
            self.bns.append(nn.BatchNorm1d(hidden))

        # MLP classification head
        self.head = nn.Sequential(
            nn.Linear(hidden, hidden//2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden//2, 1)   # logit output for BCEWithLogitsLoss
        )

    def forward(self, x, edge_index, batch):
        # Input projection
        x = F.relu(self.bn0(self.proj(x)))
        x = self.drop(x)
        # Conv layers with residual connections
        for conv, bn in zip(self.convs, self.bns):
            x_res = x
            x = conv(x, edge_index)
            x = bn(x)
            x = F.relu(x)
            x = self.drop(x)
            x = x + x_res          # skip connection (same dim)
        # Global readout + head
        x = global_mean_pool(x, batch)
        return self.head(x).squeeze(-1)   # (B,)

print(f'MolGNN defined. ATOM_DIM={ATOM_DIM}')
````

````python
# ── 18. Training & Evaluation Functions ──────────────────────────────────────
def gnn_train_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        logits = model(batch.x, batch.edge_index, batch.batch)
        loss   = criterion(logits, batch.y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
    return total_loss / len(loader.dataset)

@torch.no_grad()
def gnn_eval(model, loader, device):
    model.eval()
    all_logits, all_labels = [], []
    for batch in loader:
        batch = batch.to(device)
        all_logits.append(model(batch.x, batch.edge_index, batch.batch).cpu())
        all_labels.append(batch.y.cpu())
    logits = torch.cat(all_logits).numpy()
    labels = torch.cat(all_labels).numpy().astype(int)
    probs  = torch.sigmoid(torch.tensor(logits)).numpy()
    auc    = roc_auc_score(labels, probs)
    return probs, labels, auc

print('Training helpers defined.')
````

````python
# ── 19. GNN DataLoaders ──────────────────────────────────────────────────────
BATCH = 64
train_loader = DataLoader(graphs_train, batch_size=BATCH, shuffle=True)
val_loader   = DataLoader(graphs_val,   batch_size=BATCH, shuffle=False)
test_loader  = DataLoader(graphs_test,  batch_size=BATCH, shuffle=False)

# Positive weight for BCEWithLogitsLoss — upweights active class in GNN training
# (training set is balanced 1:1, so pos_weight=1)
pos_weight = torch.tensor([(y_train==0).sum()/(y_train==1).sum()], device=DEVICE)
criterion  = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

print(f'Train graphs: {len(graphs_train)}')
print(f'Val   graphs: {len(graphs_val)}')
print(f'Test  graphs: {len(graphs_test)}')
print(f'pos_weight   : {pos_weight.item():.3f} (1.0 = balanced)')
````

````python
# ── 20. Hyperparameter Grid & Training Loop ──────────────────────────────────
GNN_CONFIGS = {
    'GCN' : [
        {'hidden':128, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':256, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':128, 'n_layers':4, 'dropout':0.4, 'lr':0.0005},
    ],
    'GAT' : [
        {'hidden':128, 'n_layers':3, 'dropout':0.3, 'lr':0.001,  'heads':4},
        {'hidden':128, 'n_layers':3, 'dropout':0.4, 'lr':0.0005, 'heads':8},
        {'hidden':256, 'n_layers':3, 'dropout':0.3, 'lr':0.001,  'heads':4},
    ],
    'GIN' : [
        {'hidden':128, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':256, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':128, 'n_layers':4, 'dropout':0.4, 'lr':0.0005},
    ],
    'SAGE': [
        {'hidden':128, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':256, 'n_layers':3, 'dropout':0.3, 'lr':0.001},
        {'hidden':128, 'n_layers':4, 'dropout':0.4, 'lr':0.0005},
    ],
}

MAX_EPOCHS = 100
PATIENCE   = 20   # early stopping

gnn_best_models  = {}   # best state_dict per architecture
gnn_best_configs = {}   # best hyperparams per architecture
gnn_best_val_auc = {}
gnn_history      = {}   # training curves

for arch, configs in GNN_CONFIGS.items():
    print(f'\n{"="*55}')
    print(f'Architecture: {arch}  ({len(configs)} configs)')
    best_auc, best_state, best_cfg = 0, None, {}

    for cfg in configs:
        heads = cfg.pop('heads', 4)
        model = MolGNN(ATOM_DIM, hidden=cfg['hidden'], n_layers=cfg['n_layers'],
                       dropout=cfg['dropout'], conv_type=arch, heads=heads).to(DEVICE)
        opt   = torch.optim.Adam(model.parameters(), lr=cfg['lr'], weight_decay=1e-5)
        sched = torch.optim.lr_scheduler.ReduceLROnPlateau(
                    opt, mode='max', factor=0.5, patience=10)
        cfg['heads'] = heads  # restore

        best_ep_auc, no_improv = 0, 0
        best_ep_state = None
        train_aucs, val_aucs = [], []

        for epoch in range(1, MAX_EPOCHS+1):
            loss = gnn_train_epoch(model, train_loader, opt, criterion, DEVICE)
            _,_, tr_auc = gnn_eval(model, train_loader, DEVICE)
            _,_, va_auc = gnn_eval(model, val_loader,   DEVICE)
            sched.step(va_auc)
            train_aucs.append(tr_auc); val_aucs.append(va_auc)
            if va_auc > best_ep_auc:
                best_ep_auc  = va_auc
                best_ep_state = copy.deepcopy(model.state_dict())
                no_improv = 0
            else:
                no_improv += 1
            if no_improv >= PATIENCE:
                break

        if best_ep_auc > best_auc:
            best_auc   = best_ep_auc
            best_state = best_ep_state
            best_cfg   = cfg.copy()
            gnn_history[arch] = (train_aucs, val_aucs)

        cfg_str = {k:v for k,v in cfg.items()}
        print(f'  {cfg_str}  →  val AUC={best_ep_auc:.4f}  stopped@ep{len(train_aucs)}')

    gnn_best_val_auc[arch]  = best_auc
    gnn_best_configs[arch]  = best_cfg
    gnn_best_models[arch]   = best_state
    print(f'>> Best {arch}: AUC={best_auc:.4f}  cfg={best_cfg}')
````

````python
# ── 32. Fragment Attribution B: GNN Gradient Saliency — Which Atoms Matter? ──
# Gradient saliency (Simonyan et al.): compute d(prediction)/d(atom_features).
# Atoms with large gradients are the ones the GNN relied on most.
# This is the simplest explainability method — start here before GNNExplainer.

import torch
from rdkit.Chem import Draw
from rdkit.Chem.Draw import rdMolDraw2D
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
import numpy as np

# ── Load best GNN ─────────────────────────────────────────────────────────────
# Pick whichever arch had the highest test ROC-AUC
arch_order = ['GAT', 'GIN', 'SAGE', 'GCN']
best_arch  = next((a for a in arch_order if a in gnn_best_models), list(gnn_best_models.keys())[0])
cfg        = gnn_best_configs[best_arch]
explain_model = MolGNN(ATOM_DIM, hidden=cfg['hidden'], n_layers=cfg['n_layers'],
                        dropout=0.0,          # no dropout during explanation
                        conv_type=best_arch, heads=cfg.get('heads', 4)).to(DEVICE)
explain_model.load_state_dict(gnn_best_models[best_arch])
explain_model.eval()
print(f"Explaining: {best_arch}")

# ── Gradient saliency function ────────────────────────────────────────────────
def gradient_saliency(model, pyg_data, device):
    """
    Forward pass with x requiring grad → backward from prediction logit.
    Returns per-atom importance = sum of |grad| across atom feature dims.
    """
    data  = pyg_data.to(device)
    x     = data.x.clone().detach().requires_grad_(True)
    batch = torch.zeros(x.size(0), dtype=torch.long, device=device)
    logit = model(x, data.edge_index, batch)          # scalar logit
    logit.backward()
    saliency = x.grad.abs().sum(dim=1).cpu().numpy()  # (n_atoms,)
    return saliency

# ── Draw atoms coloured by saliency (white→red) ───────────────────────────────
def draw_mol_saliency(smi, saliency, size=(280, 220)):
    """Returns a PIL Image with atoms coloured by importance score."""
    mol = Chem.MolFromSmiles(str(smi))
    if mol is None:
        return Image.new('RGB', size, (220, 220, 220))
    sal = np.array(saliency, dtype=float)
    if sal.max() > 0:
        sal = sal / sal.max()
    # white (0) → red (1)
    atom_colors  = {i: (1.0, 1.0 - float(v), 1.0 - float(v)) for i, v in enumerate(sal)}
    highlight_r  = {i: 0.45 for i in range(mol.GetNumAtoms())}
    try:
        drawer = rdMolDraw2D.MolDraw2DCairo(*size)
    except AttributeError:
        drawer = rdMolDraw2D.MolDraw2DSVG(*size)
    drawer.drawOptions().addAtomIndices = False
    rdMolDraw2D.PrepareAndDrawMolecule(
        drawer, mol,
        highlightAtoms=list(range(mol.GetNumAtoms())),
        highlightAtomColors=atom_colors,
        highlightAtomRadii=highlight_r)
    drawer.FinishDrawing()
    bio = BytesIO(drawer.GetDrawingText())
    return Image.open(bio).copy()

# ── Pick compounds: 3 most-toxic + 3 most-safe predictions ────────────────────
arch_probs  = gnn_probs_test[best_arch]
sorted_idx  = np.argsort(arch_probs)
toxic_idx   = list(sorted_idx[-3:][::-1])   # top 3 highest predicted probability
safe_idx    = list(sorted_idx[:3])           # top 3 lowest predicted probability
show_idx    = toxic_idx + safe_idx
show_group  = ['Predicted Toxic'] * 3 + ['Predicted Safe'] * 3

# ── Plot ───────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(13, 9))
fig.suptitle(f"{best_arch} Gradient Saliency — Red atoms drove the prediction\n"
             "(brighter red = atom contributed more)", fontsize=12, fontweight='bold')

for ax, idx, group in zip(axes.flat, show_idx, show_group):
    graph = graphs_test[idx]
    smi   = df_test['PUBCHEM_EXT_DATASOURCE_SMILES'].iloc[idx]
    prob  = arch_probs[idx]
    true  = y_test[idx]
    sal   = gradient_saliency(explain_model, graph, DEVICE)
    img   = draw_mol_saliency(smi, sal)
    ax.imshow(img)
    ax.set_title(f"{group}\np={prob:.3f}  true={'Toxic' if true==1 else 'Safe'}",
                 fontsize=9, linespacing=1.5)
    ax.axis('off')

plt.tight_layout()
plt.savefig('fragment_saliency_gnn.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved → fragment_saliency_gnn.png")
print("\nNext step: fragments that light up red in BOTH toxic AND safe compounds are")
print("structural context — fragments unique to toxic compounds are the real alerts.")
````

- **Negative SHAP value** for a bit => that substructure *pushed toward safe* (protective fragment)
- SHAP values sum exactly to `prediction - baseline`, so there is no information lost

We use `shap.TreeExplainer`

````python
# -- 34. SHAP: XGBoost Toxicophore Attribution --------------------------------
# XGBoost is a tree ensemble -- shap.TreeExplainer gives exact, fast SHAP values.
# Each SHAP value tells us how much one bit pushed the prediction up (toxic) or down (safe).

import shap, numpy as np, matplotlib.pyplot as plt

explainer = shap.TreeExplainer(best_xgb)
shap_vals = explainer.shap_values(X_test)   # shape: (n_test, 2048)

print(f'SHAP matrix : {shap_vals.shape}')
print(f'Baseline    : {explainer.expected_value:.4f}')

# Beeswarm: top 20 bits across all test compounds
# Red dot  = bit IS set in that molecule  -> how much it pushed the prediction
# Blue dot = bit NOT set
# Rightward = pushed toward ACTIVE/TOXIC
shap.summary_plot(
    shap_vals, X_test,
    feature_names=[f'Bit {i}' for i in range(2048)],
    max_display=20, show=False, plot_type='dot'
)
plt.title('SHAP Beeswarm (XGBoost) -- Top 20 ECFP4 toxicophore bits\n'
          'Red=bit present | Rightward=pushes toward TOXIC', pad=10)
plt.tight_layout()
plt.savefig('shap_beeswarm_mito.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved => shap_beeswarm_mito.png')

# Mean |SHAP| per bit -- persistent toxicophores
mean_abs_shap = np.abs(shap_vals).mean(axis=0)   # (2048,)
top20 = np.argsort(mean_abs_shap)[::-1][:20]

fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(range(20), mean_abs_shap[top20], color='#d62728')
ax.set_xticks(range(20))
ax.set_xticklabels([f'Bit {b}' for b in top20], rotation=45, ha='right', fontsize=8)
ax.set_ylabel('Mean |SHAP value|')
ax.set_title('Top 20 persistent toxicophore bits (mean |SHAP| across test set)')
plt.tight_layout()
plt.savefig('shap_top20_mito.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved => shap_top20_mito.png')
````

## Step 13 -- GNNExplainer: Atom-level Toxicophore Attribution

SHAP (Step 12) explains the **fingerprint-based** models -- it tells us which ECFP4 bits matter.
But fingerprints lose structural context (which atoms are connected to what).

**GNNExplainer** (Ying et al. 2019) works directly on the molecular graph:
- It learns a soft mask over **atoms** (nodes) and **bonds** (edges)
- The mask is optimised so that the masked subgraph alone reproduces the model's prediction
- High mask value on an atom = that atom is essential for the toxic prediction

This gives atom-resolution toxicophores: we can highlight exactly which atoms/bonds the GNN relied on, drawn directly on the 2D structure.

````python
# -- 36. GNNExplainer: Atom-level Toxicophore Attribution ---------------------
import torch
import numpy as np
import matplotlib.pyplot as plt
from rdkit import Chem
from rdkit.Chem import Draw
from PIL import Image
from torch_geometric.explain import Explainer, GNNExplainer

# -- Load best GNN ------------------------------------------------------------
arch_order    = ['GAT', 'GIN', 'SAGE', 'GCN']
best_arch     = next((a for a in arch_order if a in gnn_best_models), list(gnn_best_models.keys())[0])
cfg           = gnn_best_configs[best_arch]
explain_model = MolGNN(ATOM_DIM, hidden=cfg['hidden'], n_layers=cfg['n_layers'],
                        dropout=0.0, conv_type=best_arch,
                        heads=cfg.get('heads', 4)).to(DEVICE)
explain_model.load_state_dict(gnn_best_models[best_arch])
explain_model.eval()
print(f'Loaded: {best_arch}')

# -- Wrapper: makes batch optional so PyG Explainer can call forward ----------
class WrappedGNN(torch.nn.Module):
    def __init__(self, m, dev):
        super().__init__()
        self.m = m; self.dev = dev
    def forward(self, x, edge_index, batch=None):
        if batch is None:
            batch = torch.zeros(x.size(0), dtype=torch.long, device=self.dev)
        return self.m(x, edge_index, batch)

wrapped = WrappedGNN(explain_model, DEVICE).to(DEVICE)
wrapped.eval()

# -- Build Explainer ----------------------------------------------------------
gnn_explainer = Explainer(
    model=wrapped,
    algorithm=GNNExplainer(epochs=200),
    explanation_type='model',
    node_mask_type='common_attributes',
    edge_mask_type='object',
    model_config=dict(mode='binary_classification', task_level='graph', return_type='raw'),
)

# -- Get top-5 most-predicted-toxic test compounds ----------------------------
gnn_probs_test = torch.zeros(len(graphs_test))
with torch.no_grad():
    for idx, g in enumerate(graphs_test):
        g2  = g.to(DEVICE)
        bat = torch.zeros(g2.x.size(0), dtype=torch.long, device=DEVICE)
        gnn_probs_test[idx] = torch.sigmoid(explain_model(g2.x, g2.edge_index, bat)).item()
top5_idx = torch.topk(gnn_probs_test, 5).indices.tolist()
print(f'Top-5 indices: {top5_idx}')

# -- Run GNNExplainer and draw simple highlights ------------------------------
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for plot_i, test_i in enumerate(top5_idx):
    g      = graphs_test[test_i].to(DEVICE)
    prob   = float(gnn_probs_test[test_i])
    true   = int(y_test[test_i])
    smiles = df_test.iloc[test_i]['csmi']
    mol    = Chem.MolFromSmiles(str(smiles))

    exp        = gnn_explainer(x=g.x, edge_index=g.edge_index)
    scores     = exp.node_mask.detach().cpu().numpy().flatten()   # (n_graph_atoms,)
    n_mol      = mol.GetNumAtoms()

    # Normalise and pick top-5 atoms that are valid in the RDKit mol
    scores_norm = scores / (scores.max() + 1e-8)
    ranked      = np.argsort(scores_norm)[::-1]
    top_atoms   = [int(a) for a in ranked if int(a) < n_mol][:5]   # guard out-of-range

    # Simple red highlight on top atoms
    img = Draw.MolToImage(
        mol,
        size=(300, 250),
        highlightAtoms=top_atoms,
        highlightColor=(0.9, 0.2, 0.2),
    )

    lbl = 'ACT' if true == 1 else 'INA'
    axes[plot_i].imshow(img)
    axes[plot_i].set_title(f'#{test_i} {lbl} P={prob:.2f}\ntop atoms: {top_atoms}', fontsize=8)
    axes[plot_i].axis('off')
    print(f'  [{plot_i+1}/5] #{test_i} P={prob:.3f} true={lbl} highlighted={top_atoms}')

fig.suptitle(
    f'GNNExplainer ({best_arch}) -- Red atoms drove the toxic prediction',
    fontsize=12
)
plt.tight_layout()
plt.savefig('gnnexplainer_toxicophores_mito.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved => gnnexplainer_toxicophores_mito.png')
````

````python
# -- 37. GNNExplainer: Aggregate Atom-type Importance Across Test Set ---------
# Run GNNExplainer on all correctly-predicted ACTIVE test compounds.
# For each atom, record its atomic symbol and importance score.
# This gives a population-level view: which ELEMENT types are consistently toxic.

from collections import defaultdict
import numpy as np
import matplotlib.pyplot as plt
from rdkit import Chem

# Only explain correctly-predicted actives (true positives)
tp_indices = [
    i for i in range(len(graphs_test))
    if y_test[i] == 1 and gnn_probs_test[i] >= 0.5
]
print(f'True positives in test set: {len(tp_indices)}')
print('Running GNNExplainer on all TPs (may take a few minutes)...')

atom_importance = defaultdict(list)   # symbol -> list of importance scores

for idx in tp_indices:
    g      = graphs_test[idx].to(DEVICE)
    smiles = df_test.iloc[idx]['csmi']
    mol    = Chem.MolFromSmiles(str(smiles))
    if mol is None: continue

    explanation = gnn_explainer(x=g.x, edge_index=g.edge_index)
    node_mask   = explanation.node_mask.detach().cpu().numpy().squeeze()
    # normalise per-molecule so all compounds contribute equally
    node_mask   = node_mask / (node_mask.sum() + 1e-8)

    for atom_i, atom in enumerate(mol.GetAtoms()):
        if atom_i >= len(node_mask): break   # guard: GNN graph may have fewer atoms than RDKit mol
        symbol = atom.GetSymbol()
        atom_importance[symbol].append(float(node_mask[atom_i]))

# Mean importance per element
elements  = sorted(atom_importance.keys(), key=lambda s: -np.mean(atom_importance[s]))
mean_imp  = [np.mean(atom_importance[e]) for e in elements]
count_imp = [len(atom_importance[e])     for e in elements]

# Plot only elements with >= 5 observations
mask_enough = [c >= 5 for c in count_imp]
elements_f  = [e for e, m in zip(elements, mask_enough) if m]
mean_imp_f  = [v for v, m in zip(mean_imp, mask_enough) if m]

fig, ax = plt.subplots(figsize=(10, 4))
bars = ax.bar(elements_f, mean_imp_f, color='#d62728', alpha=0.85)
ax.set_xlabel('Atom element')
ax.set_ylabel('Mean normalised GNNExplainer importance')
ax.set_title(
    f'GNNExplainer ({best_arch}) -- Which atom types drive mito-toxicity?\n'
    f'Aggregated across {len(tp_indices)} true-positive compounds'
)
plt.tight_layout()
plt.savefig('gnnexplainer_atom_elements_mito.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved => gnnexplainer_atom_elements_mito.png')

print('\nElement importance summary:')
print(f'{"Element":>8}  {"Mean imp":>10}  {"Count":>7}')
for e, m, c in zip(elements_f, mean_imp_f, [count_imp[i] for i, ok in enumerate(mask_enough) if ok]):
    print(f'{e:>8}  {m:>10.6f}  {c:>7}')
````
