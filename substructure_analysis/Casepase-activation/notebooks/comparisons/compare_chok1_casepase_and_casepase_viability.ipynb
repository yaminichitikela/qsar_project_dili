# Compare Caspase Activation (AID 1346980) and Caspase Viability (AID 1346979) - CHO-K1

This notebook compares two Caspase-3/7 datasets in CHO-K1 cell line:
- **AID 1346980**: Caspase-3/7 Activation CHO-K1
- **AID 1346979**: Caspase-3/7 Viability Counter Screen CHO-K1

**Four Cases Analysis:**
1. Case 1: Both Caspase Activation and Caspase Viability Active
2. Case 2: Caspase Activation Active, Caspase Viability Inactive
3. Case 3: Caspase Activation Inactive, Caspase Viability Active
4. Case 4: Both Caspase Activation and Caspase Viability Inactive

```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

## Step 1: Load AID 1346980 (Caspase Activation CHO-K1) and AID 1346979 (Caspase Viability CHO-K1)

```python
# Load AID 1346980 - Caspase-3/7 Activation CHO-K1
aid_1346980_caspase = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_1346980_cleaned.csv', 
                        low_memory=False)

# Load AID 1346979 - Caspase-3/7 Viability Counter Screen CHO-K1
aid_1346979_viability = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_1346979_cleaned.csv', 
                        low_memory=False)

print(pd.DataFrame(aid_1346979_viability).head(5))

print("AID 1346980 - Caspase-3/7 Activation CHO-K1")
print(f"Loaded shape: {aid_1346980_caspase.shape}")
print(f"Unique CIDs: {aid_1346980_caspase['PUBCHEM_CID'].nunique()}")

print("\nAID 1346979 - Caspase-3/7 Viability CHO-K1")
print(f"Loaded shape: {aid_1346979_viability.shape}")
print(f"Unique CIDs: {aid_1346979_viability['PUBCHEM_CID'].nunique()}")
```

```python
# Sort by CID and activity score (highest first)
aid_1346980_sorted = aid_1346980_caspase.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)
aid_1346979_sorted = aid_1346979_viability.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

# CASE 1: Both Caspase Activation and Caspase Viability Active
filtered_1346980_active = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_1346979_active = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case1_common_cids = set(filtered_1346980_active['PUBCHEM_CID']).intersection(set(filtered_1346979_active['PUBCHEM_CID']))
print(f"Case 1 - Both Active: {len(case1_common_cids)} compounds")
```

```python
# Get rows for Case 1 compounds
case1_1346980 = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_CID'].isin(case1_common_cids)] 
case1_1346979 = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_CID'].isin(case1_common_cids)]

# Combine Case 1
case1_merged = pd.concat([case1_1346980, case1_1346979], ignore_index=True)
```

```python
# Save Case 1: Both Caspase Activation and Viability Active
case1_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/chok1_comparisons/chok1_casepase_and_casepase_viability_active.csv', index=False)
```

```python
# CASE 2: Caspase Activation (1346980) Active, Caspase Viability (1346979) Inactive
filtered_1346980_active = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_1346979_inactive = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

case2_common_cids = set(filtered_1346980_active['PUBCHEM_CID']).intersection(set(filtered_1346979_inactive['PUBCHEM_CID']))
print(f"Case 2 - Caspase Active, Viability Inactive: {len(case2_common_cids)} compounds")
```

```python
# Get rows for Case 2 compounds
case2_1346980 = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_CID'].isin(case2_common_cids)] 
case2_1346979 = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_CID'].isin(case2_common_cids)]

# Combine Case 2
case2_merged = pd.concat([case2_1346980, case2_1346979], ignore_index=True)
```

```python
# Save Case 2: Caspase Activation Active, Viability Inactive
case2_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/chok1_comparisons/chok1_casepase_active_and_casepase_viability_inactive.csv', index=False)
```

```python
# CASE 3: Caspase Activation (1346980) Inactive, Caspase Viability (1346979) Active
filtered_1346980_inactive = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']
filtered_1346979_active = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case3_common_cids = set(filtered_1346980_inactive['PUBCHEM_CID']).intersection(set(filtered_1346979_active['PUBCHEM_CID']))
print(f"Case 3 - Caspase Inactive, Viability Active: {len(case3_common_cids)} compounds")
```

```python
# Get rows for Case 3 compounds
case3_1346980 = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_CID'].isin(case3_common_cids)] 
case3_1346979 = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_CID'].isin(case3_common_cids)]

# Combine Case 3
case3_merged = pd.concat([case3_1346980, case3_1346979], ignore_index=True)
```

```python
# Save Case 3: Caspase Activation Inactive, Viability Active
case3_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/chok1_comparisons/chok1_casepase_inactive_and_casepase_viability_active.csv', index=False)
```

```python
# CASE 4: Both Caspase Activation and Caspase Viability Inactive
filtered_1346980_inactive = aid_1346980_sorted[aid_1346980_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']
filtered_1346979_inactive = aid_1346979_sorted[aid_1346979_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

case4_common_cids = set(filtered_1346980_inactive['PUBCHEM_CID']).intersection(set(filtered_1346979_inactive['PUBCHEM_CID']))
print(f"Case 4 - Both Inactive: {len(case4_common_cids)} compounds")
```
