# Compare Mitochondrial Toxicity (AID 720635) and Mitochondrial Viability (AID 720634)

This notebook compares two mitochondrial datasets:
- **AID 720635**: Mitochondrial Toxicity HepG2
- **AID 720634**: Mitochondrial Viability HepG2

**Four Cases Analysis:**
1. Case 1: Both Mito Toxicity and Mito Viability Active
2. Case 2: Mito Toxicity Active, Mito Viability Inactive
3. Case 3: Mito Toxicity Inactive, Mito Viability Active
4. Case 4: Both Mito Toxicity and Mito Viability Inactive

```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

## Step 1: Load AID 720635 (Mito Toxicity) and AID 720634 (Mito Viability)

```python
# Load AID 720635 - Mitochondrial Toxicity HepG2
aid_720635_mito_tox = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720635_cleaned.csv', 
                       skiprows=[1,2,3,4,5], 
                       low_memory=False)

# Load AID 720634 - Mitochondrial Viability HepG2
aid_720634_mito_viab = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720634_cleaned.csv', 
                       skiprows=[1,2,3,4,5], 
                       low_memory=False)

print("AID 720635 - Mitochondrial Toxicity HepG2")
print(f"Loaded shape: {aid_720635_mito_tox.shape}")
print(f"Unique CIDs: {aid_720635_mito_tox['PUBCHEM_CID'].nunique()}")

print("\nAID 720634 - Mitochondrial Viability HepG2")
print(f"Loaded shape: {aid_720634_mito_viab.shape}")
print(f"Unique CIDs: {aid_720634_mito_viab['PUBCHEM_CID'].nunique()}")
```

```python
# Sort by CID and activity score (highest first)
aid_720635_sorted = aid_720635_mito_tox.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)
aid_720634_sorted = aid_720634_mito_viab.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

# CASE 1: Both Mito Toxicity and Mito Viability Active
filtered_720635_active = aid_720635_sorted[aid_720635_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_720634_active = aid_720634_sorted[aid_720634_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case1_common_cids = set(filtered_720635_active['PUBCHEM_CID']).intersection(set(filtered_720634_active['PUBCHEM_CID']))
print(f"Case 1 - Both Active: {len(case1_common_cids)} compounds")
```

```python
# Get rows for Case 1 compounds
case1_720635 = aid_720635_sorted[aid_720635_sorted['PUBCHEM_CID'].isin(case1_common_cids)] 
case1_720634 = aid_720634_sorted[aid_720634_sorted['PUBCHEM_CID'].isin(case1_common_cids)]

# Combine and save Case 1
case1_merged = pd.concat([case1_720635, case1_720634], ignore_index=True)
```

```python
# Save Case 1: Both Mito Toxicity and Mito Viability Active
case1_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/mito_viability_comparisons/mito_and_mito_viability_active.csv', index=False)
```

```python
# CASE 2: Mito Toxicity (720635) Active, Mito Viability (720634) Inactive
filtered_720635_active = aid_720635_sorted[aid_720635_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_720634_inactive = aid_720634_sorted[aid_720634_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

case2_common_cids = set(filtered_720635_active['PUBCHEM_CID']).intersection(set(filtered_720634_inactive['PUBCHEM_CID']))
print(f"Case 2 - Mito Tox Active, Mito Viab Inactive: {len(case2_common_cids)} compounds")
```

```python
# Get rows for Case 2 compounds
case2_720635 = aid_720635_sorted[aid_720635_sorted['PUBCHEM_CID'].isin(case2_common_cids)] 
case2_720634 = aid_720634_sorted[aid_720634_sorted['PUBCHEM_CID'].isin(case2_common_cids)]

# Combine and save Case 2
case2_merged = pd.concat([case2_720635, case2_720634], ignore_index=True)
```

```python
# Save Case 2: Mito Toxicity Active, Mito Viability Inactive
case2_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/mito_viability_comparisons/mito_active_and_mito_viability_inctive.csv', index=False)
```

```python
# CASE 3: Mito Toxicity (720635) Inactive, Mito Viability (720634) Active
filtered_720635_inactive = aid_720635_sorted[aid_720635_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']
filtered_720634_active = aid_720634_sorted[aid_720634_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case3_common_cids = set(filtered_720635_inactive['PUBCHEM_CID']).intersection(set(filtered_720634_active['PUBCHEM_CID']))
print(f"Case 3 - Mito Tox Inactive, Mito Viab Active: {len(case3_common_cids)} compounds")
```

```python
# Get rows for Case 3 compounds
case3_720635 = aid_720635_sorted[aid_720635_sorted['PUBCHEM_CID'].isin(case3_common_cids)] 
case3_720634 = aid_720634_sorted[aid_720634_sorted['PUBCHEM_CID'].isin(case3_common_cids)]

# Combine and save Case 3
case3_merged = pd.concat([case3_720635, case3_720634], ignore_index=True)
```

```python
# Save Case 3: Mito Toxicity Inactive, Mito Viability Active (no values found)
case3_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/mito_viability_comparisons/mito_inactive_and_viability_active.csv', index=False)
```

```python
# CASE 4: Both Mito Toxicity and Mito Viability Inactive
filtered_720635_inactive = aid_720635_sorted[aid_720635_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']
filtered_720634_inactive = aid_720634_sorted[aid_720634_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

case4_common_cids = set(filtered_720635_inactive['PUBCHEM_CID']).intersection(set(filtered_720634_inactive['PUBCHEM_CID']))
print(f"Case 4 - Both Inactive: {len(case4_common_cids)} compounds")
```
