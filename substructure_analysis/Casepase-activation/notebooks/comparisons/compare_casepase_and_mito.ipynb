# Compare Caspase (HepG2) and Mitochondrial Toxicity (HepG2)

This notebook compares two datasets:
- **AID 720635**: Mitochondrial Toxicity HepG2
- **Combined HepG2 Caspase**: Caspase-3/7 activation HepG2

**Four Cases Analysis:**
1. Case 1: Both Caspase and Mito Active
2. Case 2: Caspase Active, Mito Inactive
3. Case 3: Caspase Inactive, Mito Active
4. Case 4: Both Caspase and Mito Inactive

```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

## Step 1: Load AID 720635 (Mito Toxicity) and Combined HepG2 Caspase Data

```python
# Load AID 720635 - Mitochondrial Toxicity HepG2
aid_720635_mito = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720635_cleaned.csv', 
                       low_memory=False)

# Load Combined HepG2 Caspase data
caspase_hepg2 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/combined_hepg2_casepase.csv', 
                       low_memory=False)

print("AID 720635 - Mitochondrial Toxicity HepG2")
print(f"Loaded shape: {aid_720635_mito.shape}")
print(f"Unique CIDs: {aid_720635_mito['PUBCHEM_CID'].nunique()}")

print("\nCombined HepG2 Caspase")
print(f"Loaded shape: {caspase_hepg2.shape}")
print(f"Unique CIDs: {caspase_hepg2['PUBCHEM_CID'].nunique()}")
```

```python
# Sort by CID and activity score (highest first)
aid_720635_mito_sorted = aid_720635_mito.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)
caspase_hepg2_sorted = caspase_hepg2.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

# CASE 1: Both Caspase and Mito Active
filtered_mito_active = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_caspase_active = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case1_common_cids = set(filtered_mito_active['PUBCHEM_CID']).intersection(set(filtered_caspase_active['PUBCHEM_CID']))
print(f"Case 1 - Both Active: {len(case1_common_cids)} compounds")
```

```python
# Get rows for Case 1 compounds from both datasets
case1_mito = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_CID'].isin(case1_common_cids)] 
case1_caspase = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_CID'].isin(case1_common_cids)]

# Combine and save Case 1
case1_merged = pd.concat([case1_mito, case1_caspase], ignore_index=True)
```

```python
# Save Case 1: Both Caspase and Mito Active
case1_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/casepase_mito_comparisons/caspase_and_mito_Active.csv', index=False)
```

```python
# CASE 2: Caspase Active, Mito Inactive
filtered_mito_inactive = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']
filtered_caspase_active = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']

case2_common_cids = set(filtered_mito_inactive['PUBCHEM_CID']).intersection(set(filtered_caspase_active['PUBCHEM_CID']))
print(f"Case 2 - Caspase Active, Mito Inactive: {len(case2_common_cids)} compounds")
```

```python
# Get rows for Case 2 compounds
case2_mito = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_CID'].isin(case2_common_cids)] 
case2_caspase = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_CID'].isin(case2_common_cids)]

# Combine and save Case 2
case2_merged = pd.concat([case2_mito, case2_caspase], ignore_index=True)
```

```python
# Save Case 2: Caspase Active, Mito Inactive
case2_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/casepase_mito_comparisons/caspase_active_and_mito_inctive.csv', index=False)
```
