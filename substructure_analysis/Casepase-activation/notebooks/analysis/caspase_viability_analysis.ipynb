# Caspase & Viability Analysis

analysis of the merged caspase-viability dataset.

**Assay Types:**
- `caspase` - Caspase-3/7 assay (14 cell lines)
- `viability` - Viability assay (CHO-K1)
- `mitochondrial_viability_hepg2` - Mitochondrial viability (HepG2)
- `caspase_viability_hepg2` - Caspase viability counter-screen (HepG2)

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 200)
pd.set_option('display.max_rows', 100)

# Load the merged dataset
df = pd.read_csv('output/merged_caspase_viability_final.csv')

print(f"Dataset Shape: {df.shape}")
print(f"Unique CIDs: {df['CID'].nunique():,}")
print(f"\nAssay Types: {df['assay_type'].unique().tolist()}")
print(f"\nCell Lines ({df['cell_line'].nunique()}): {df['cell_line'].unique().tolist()}")
```

```python
# Define viability types
# viability = CHO-K1
# mitochondrial_viability_hepg2 = HepG2
# caspase_viability_hepg2 = HepG2
VIABILITY_TYPES = ['viability', 'mitochondrial_viability_hepg2', 'caspase_viability_hepg2']

print("Viability assay types used in this analysis:")
print(f"  - viability (CHO-K1)")
print(f"  - mitochondrial_viability_hepg2 (HepG2)")
print(f"  - caspase_viability_hepg2 (HepG2)")
```

---
## Block 1: Overview - Counts and Percentages

Summary of Active, Inactive, Inconclusive for Caspase and Viability assays.

```python
# CASPASE Summary
print("CASPASE ASSAY SUMMARY")

caspase_df = df[df['assay_type'] == 'caspase']

total_records = len(caspase_df)
total_cids = caspase_df['CID'].nunique()

# Record counts
active_records = len(caspase_df[caspase_df['outcome'] == 'Active'])
inactive_records = len(caspase_df[caspase_df['outcome'] == 'Inactive'])
inconclusive_records = len(caspase_df[caspase_df['outcome'] == 'Inconclusive'])

# Unique CID counts for each outcome
active_cids = caspase_df[caspase_df['outcome'] == 'Active']['CID'].nunique()
inactive_cids = caspase_df[caspase_df['outcome'] == 'Inactive']['CID'].nunique()
inconclusive_cids = caspase_df[caspase_df['outcome'] == 'Inconclusive']['CID'].nunique()

print(f"\nTotal Records: {total_records:,}")
print(f"Total Unique CIDs: {total_cids:,}")

print(f"\n{'Outcome':<15} {'Records':>10} {'Record %':>10} {'Unique CIDs':>12} {'CID %':>10}")
print("-"*60)
print(f"{'Active':<15} {active_records:>10,} {active_records/total_records*100:>9.2f}% {active_cids:>12,} {active_cids/total_cids*100:>9.2f}%")
print(f"{'Inactive':<15} {inactive_records:>10,} {inactive_records/total_records*100:>9.2f}% {inactive_cids:>12,} {inactive_cids/total_cids*100:>9.2f}%")
print(f"{'Inconclusive':<15} {inconclusive_records:>10,} {inconclusive_records/total_records*100:>9.2f}% {inconclusive_cids:>12,} {inconclusive_cids/total_cids*100:>9.2f}%")
print("-"*60)
```

```python
# VIABILITY Summary (all 3 viability types)
print("VIABILITY ASSAY SUMMARY")
print("(viability CHO-K1 + mitochondrial_viability_hepg2 + caspase_viability_hepg2)")

viability_df = df[df['assay_type'].isin(VIABILITY_TYPES)]

total_records = len(viability_df)
total_cids = viability_df['CID'].nunique()

# Record counts
active_records = len(viability_df[viability_df['outcome'] == 'Active'])
inactive_records = len(viability_df[viability_df['outcome'] == 'Inactive'])
inconclusive_records = len(viability_df[viability_df['outcome'] == 'Inconclusive'])

# Unique CID counts for each outcome
active_cids = viability_df[viability_df['outcome'] == 'Active']['CID'].nunique()
inactive_cids = viability_df[viability_df['outcome'] == 'Inactive']['CID'].nunique()
inconclusive_cids = viability_df[viability_df['outcome'] == 'Inconclusive']['CID'].nunique()

print(f"\nTotal Records: {total_records:,}")
print(f"Total Unique CIDs: {total_cids:,}")

print(f"\n{'Outcome':<15} {'Records':>10} {'Record %':>10} {'Unique CIDs':>12} {'CID %':>10}")
print("-"*60)
print(f"{'Active':<15} {active_records:>10,} {active_records/total_records*100:>9.2f}% {active_cids:>12,} {active_cids/total_cids*100:>9.2f}%")
print(f"{'Inactive':<15} {inactive_records:>10,} {inactive_records/total_records*100:>9.2f}% {inactive_cids:>12,} {inactive_cids/total_cids*100:>9.2f}%")
print(f"{'Inconclusive':<15} {inconclusive_records:>10,} {inconclusive_records/total_records*100:>9.2f}% {inconclusive_cids:>12,} {inconclusive_cids/total_cids*100:>9.2f}%")
print("-"*60)
```

```python
# VIABILITY Breakdown by Type
print("VIABILITY BREAKDOWN BY ASSAY TYPE")

for vtype in VIABILITY_TYPES:
    vdf = df[df['assay_type'] == vtype]
    total = len(vdf)
    if total == 0:
        continue
    cell_line = vdf['cell_line'].unique()[0]
    active = len(vdf[vdf['outcome'] == 'Active'])
    inactive = len(vdf[vdf['outcome'] == 'Inactive'])
    inconclusive = len(vdf[vdf['outcome'] == 'Inconclusive'])
    
    print(f"\n--- {vtype} ({cell_line}) ---")
    print(f"Total: {total:,} | Active: {active:,} ({active/total*100:.1f}%) | Inactive: {inactive:,} ({inactive/total*100:.1f}%) | Inconclusive: {inconclusive:,} ({inconclusive/total*100:.1f}%)")
```

---
## Block 2: Cell Line Summary Table

For each cell line: Total Tested, Active, Inactive, Inconclusive, Active Rate

```python
# CASPASE by Cell Line
print("CASPASE BY CELL LINE")

caspase_df = df[df['assay_type'] == 'caspase']

caspase_cellline = []
for cell in sorted(caspase_df['cell_line'].unique()):
    cell_data = caspase_df[caspase_df['cell_line'] == cell]
    total = len(cell_data)
    active = len(cell_data[cell_data['outcome'] == 'Active'])
    inactive = len(cell_data[cell_data['outcome'] == 'Inactive'])
    inconclusive = len(cell_data[cell_data['outcome'] == 'Inconclusive'])
    
    caspase_cellline.append({
        'Cell Line': cell,
        'Total Tested': total,
        'Active': active,
        'Inactive': inactive,
        'Inconclusive': inconclusive,
        'Active Rate (%)': round(active/total*100, 2) if total > 0 else 0
    })

caspase_cellline_df = pd.DataFrame(caspase_cellline)
caspase_cellline_df = caspase_cellline_df.sort_values('Total Tested', ascending=False)
print(caspase_cellline_df.to_string(index=False))
```

```python
# VIABILITY by Cell Line
print("VIABILITY BY CELL LINE")
print("(viability + mitochondrial_viability_hepg2 + caspase_viability_hepg2)")

viability_df = df[df['assay_type'].isin(VIABILITY_TYPES)]

viability_cellline = []
for cell in sorted(viability_df['cell_line'].unique()):
    cell_data = viability_df[viability_df['cell_line'] == cell]
    total = len(cell_data)
    active = len(cell_data[cell_data['outcome'] == 'Active'])
    inactive = len(cell_data[cell_data['outcome'] == 'Inactive'])
    inconclusive = len(cell_data[cell_data['outcome'] == 'Inconclusive'])
    
    viability_cellline.append({
        'Cell Line': cell,
        'Total Tested': total,
        'Active': active,
        'Inactive': inactive,
        'Inconclusive': inconclusive,
        'Active Rate (%)': round(active/total*100, 2) if total > 0 else 0
    })

viability_cellline_df = pd.DataFrame(viability_cellline)
viability_cellline_df = viability_cellline_df.sort_values('Total Tested', ascending=False)
print(viability_cellline_df.to_string(index=False))
```

---
## Block 3: CIDs with BOTH Caspase=Active AND Viability=Active

Find CIDs where:
- Caspase assay shows **Active**
- Viability assay

```python
# Step 1: Get CIDs where Caspase = Active
caspase_active_cids = df[
    (df['assay_type'] == 'caspase') & 
    (df['outcome'] == 'Active')
]['CID'].unique()

print(f"CIDs with Caspase = Active: {len(caspase_active_cids):,}")

# Step 2: Get CIDs where Viability = Active (viability or mitochondrial_viability_hepg2)
viability_active_cids = df[
    (df['assay_type'].isin(VIABILITY_TYPES)) & 
    (df['outcome'] == 'Active')
]['CID'].unique()

print(f"CIDs with Viability = Active: {len(viability_active_cids):,}")

# Step 3: Find intersection - CIDs with BOTH Active
both_active_cids = set(caspase_active_cids).intersection(set(viability_active_cids))

print(f"CIDs with BOTH Caspase=Active AND Viability=Active: {len(both_active_cids):,}")
```

```python
# Show these CIDs
print(f"List of {len(both_active_cids)} CIDs with both Caspase=Active AND Viability=Active:")
print(sorted(both_active_cids))
```

```python
# Show details for these CIDs
both_active_details = df[
    (df['CID'].isin(both_active_cids)) & 
    (df['outcome'] == 'Active') &
    (df['assay_type'].isin(['caspase'] + VIABILITY_TYPES))
][['CID', 'assay_type', 'cell_line', 'outcome', 'activity_score']].sort_values(['CID', 'assay_type'])

print(f"\nDetails of Active records for these {len(both_active_cids)} CIDs:")
print(both_active_details.to_string(index=False))
```

---
## Block 4: Both Caspase & Viability Active - By Cell Line

Table showing for each cell line how many CIDs have both Caspase=Active and Viability=Active

```python
# Analysis by Cell Line - Both Active (SAME CELL LINE)
print("="*70)
print("CASPASE=ACTIVE AND VIABILITY=ACTIVE BY CELL LINE (SAME CELL LINE)")
print("="*70)
print("Note: Viability data exists only for HepG2 and CHO-K1")
print()

# Get caspase cell lines
caspase_df = df[df['assay_type'] == 'caspase']
viability_df = df[df['assay_type'].isin(VIABILITY_TYPES)]

caspase_cell_lines = sorted(caspase_df['cell_line'].unique())

results = []
for cell in caspase_cell_lines:
    # Total CIDs tested for caspase in this cell line
    caspase_in_cell = caspase_df[caspase_df['cell_line'] == cell]
    total_caspase_tested = caspase_in_cell['CID'].nunique()
    
    # Caspase Active in this cell line
    casp_active = set(caspase_in_cell[caspase_in_cell['outcome'] == 'Active']['CID'].unique())
    
    # Viability Active in THIS SAME cell line
    viability_in_cell = viability_df[viability_df['cell_line'] == cell]
    total_viability_tested = viability_in_cell['CID'].nunique()
    viab_active = set(viability_in_cell[viability_in_cell['outcome'] == 'Active']['CID'].unique())
    
    # Both Caspase Active AND Viability Active in SAME cell line
    both_active = casp_active.intersection(viab_active)
    
    results.append({
        'Cell Line': cell,
        'Caspase Tested': total_caspase_tested,
        'Caspase Active': len(casp_active),
        'Viability Tested': total_viability_tested,
        'Viability Active': len(viab_active),
        'Both Active': len(both_active),
        'Both Rate (%)': round(len(both_active)/len(casp_active)*100, 2) if len(casp_active) > 0 else 0
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values('Caspase Active', ascending=False)
print(results_df.to_string(index=False))

# Summary for cell lines WITH viability data
print(f"\n" + "-"*70)
hepg2_both = results_df[results_df['Cell Line'] == 'HepG2']['Both Active'].values[0]
chok1_both = results_df[results_df['Cell Line'] == 'CHO-K1']['Both Active'].values[0]
print(f"HepG2: {hepg2_both} CIDs with both Caspase=Active AND Viability=Active")
print(f"CHO-K1: {chok1_both} CIDs with both Caspase=Active AND Viability=Active")
```

```python
# HepG2 Specific: Both Active IN THE SAME CELL LINE
print("="*70)
print("HepG2: BOTH CASPASE=ACTIVE AND VIABILITY=ACTIVE (SAME CELL LINE)")
print("="*70)

# Caspase Active in HepG2
caspase_active_hepg2 = set(df[
    (df['assay_type'] == 'caspase') & 
    (df['cell_line'] == 'HepG2') & 
    (df['outcome'] == 'Active')
]['CID'].unique())

# Viability Active in HepG2 (mitochondrial_viability_hepg2 + caspase_viability_hepg2)
viability_active_hepg2 = set(df[
    (df['assay_type'].isin(['mitochondrial_viability_hepg2', 'caspase_viability_hepg2'])) & 
    (df['cell_line'] == 'HepG2') & 
    (df['outcome'] == 'Active')
]['CID'].unique())

both_active_hepg2 = caspase_active_hepg2.intersection(viability_active_hepg2)

print(f"\nCaspase Active in HepG2: {len(caspase_active_hepg2):,}")
print(f"Viability Active in HepG2 (mito + caspase_viability): {len(viability_active_hepg2):,}")
print(f"\n>>> BOTH Active in HepG2: {len(both_active_hepg2):,}")

if len(both_active_hepg2) > 0:
    print(f"\nCIDs: {sorted(both_active_hepg2)}")
```

```python
# CHO-K1 Specific: Both Active IN THE SAME CELL LINE
print("CHO-K1: BOTH CASPASE=ACTIVE AND VIABILITY=ACTIVE (SAME CELL LINE)")

# Caspase Active in CHO-K1
caspase_active_chok1 = set(df[
    (df['assay_type'] == 'caspase') & 
    (df['cell_line'] == 'CHO-K1') & 
    (df['outcome'] == 'Active')
]['CID'].unique())

# Viability Active in CHO-K1
viability_active_chok1 = set(df[
    (df['assay_type'] == 'viability') & 
    (df['cell_line'] == 'CHO-K1') & 
    (df['outcome'] == 'Active')
]['CID'].unique())

both_active_chok1 = caspase_active_chok1.intersection(viability_active_chok1)

print(f"\nCaspase Active in CHO-K1: {len(caspase_active_chok1):,}")
print(f"Viability Active in CHO-K1: {len(viability_active_chok1):,}")
print(f"\n>>> BOTH Active in CHO-K1: {len(both_active_chok1):,}")

if len(both_active_chok1) > 0:
    print(f"\nCIDs: {sorted(both_active_chok1)}")
```

##  Export Data - Case 1: Caspase Active & Viability Active


```python
# Case 1: Export CIDs where BOTH Caspase=Active AND Viability=Active
print("="*70)
print("CASE 1: EXPORTING CASPASE ACTIVE & VIABILITY ACTIVE DATA")
print("="*70)

# Get all CIDs with both caspase active and viability active
caspase_active_cids = set(df[
    (df['assay_type'] == 'caspase') & 
    (df['outcome'] == 'Active')
]['CID'].unique())

viability_active_cids = set(df[
    (df['assay_type'].isin(VIABILITY_TYPES)) & 
    (df['outcome'] == 'Active')
]['CID'].unique())

case1_cids = caspase_active_cids.intersection(viability_active_cids)

print(f"\nTotal CIDs with both Caspase=Active AND Viability=Active: {len(case1_cids)}")

# Get all records for these CIDs (Active outcomes only for caspase and viability)
case1_data = df[
    (df['CID'].isin(case1_cids)) & 
    (df['outcome'] == 'Active') &
    (df['assay_type'].isin(['caspase'] + VIABILITY_TYPES))
][['CID', 'SMILES', 'assay_type', 'cell_line', 'outcome', 'activity_score', 'AID', 'assay_name']].copy()

# Sort by CID for easy verification
case1_data = case1_data.sort_values(['CID', 'cell_line', 'assay_type'])

print(f"Total records: {len(case1_data)}")
print(f"\nPreview:")
print(case1_data.head(20).to_string(index=False))

# Save to CSV
output_path = 'output/case1_caspase_viability_both_active.csv'
case1_data.to_csv(output_path, index=False)
print(f"\n>>> Saved to: {output_path}")
print(f"    Shape: {case1_data.shape}")
```

```python
# Create summary table grouped by CID for easy count verification
print("CASE 1: SUMMARY BY CID (for count verification)")

case1_summary = case1_data.groupby('CID').agg({
    'assay_type': lambda x: ', '.join(sorted(x.unique())),
    'cell_line': lambda x: ', '.join(sorted(x.unique())),
    'activity_score': 'max',
    'SMILES': 'first'
}).reset_index()

case1_summary.columns = ['CID', 'Active_Assay_Types', 'Active_Cell_Lines', 'Max_Activity_Score', 'SMILES']
case1_summary = case1_summary.sort_values('CID')

print(f"\nTotal unique CIDs: {len(case1_summary)}")
print(f"\nSummary table (first 30 rows):")
print(case1_summary.head(30).to_string(index=False))

# Save summary to CSV
summary_path = 'output/case1_summary_by_cid.csv'
case1_summary.to_csv(summary_path, index=False)
print(f"\n>>> Summary saved to: {summary_path}")
print(f"    Total CIDs: {len(case1_summary)}")
```
