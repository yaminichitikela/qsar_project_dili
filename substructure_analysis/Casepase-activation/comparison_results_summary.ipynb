# Comparison Results Summary
Simple overview of counts from all comparison datasets

```python
import pandas as pd
import os
from pathlib import Path
```

```python
results_dir = Path('results')
```

## CHO-K1 Comparisons

```python
chok1_dir = results_dir / 'chok1_comparisons'
chok1_files = {
    'Both Active': 'chok1_casepase_and_casepase_viability_active.csv',
    'Caspase Active, Viability Inactive': 'chok1_casepase_active_and_casepase_viability_inactive.csv',
    'Caspase Inactive, Viability Active': 'chok1_casepase_inactive_and_casepase_viability_active.csv',
    'Both Inactive': 'chok1_casepase_inactive_and_casepase_viability_inactive.csv'
}

chok1_counts = {}
chok1_unique_cids = {}
for label, filename in chok1_files.items():
    filepath = chok1_dir / filename
    if filepath.exists():
        df = pd.read_csv(filepath, low_memory=False)
        chok1_counts[label] = len(df)
        chok1_unique_cids[label] = df['PUBCHEM_CID'].nunique()

print("CHO-K1 Comparison Counts:")
print("-" * 50)
for label in chok1_counts:
    print(f"{label}: {chok1_counts[label]} rows, {chok1_unique_cids[label]} unique CIDs")
    
print(f"\nTotal: {sum(chok1_counts.values())} rows, {sum(chok1_unique_cids.values())} unique CIDs")
```

## Export All Summary Results to Excel

```python
# Create DataFrames for each comparison and export to Excel

# 1. CHO-K1 Comparisons
chok1_df = pd.DataFrame({
    'Comparison': list(chok1_counts.keys()),
    'Rows': list(chok1_counts.values()),
    'Unique CIDs': list(chok1_unique_cids.values())
})
chok1_df.loc[len(chok1_df)] = ['TOTAL', sum(chok1_counts.values()), sum(chok1_unique_cids.values())]

# 2. HepG2 Comparisons
hepg2_df = pd.DataFrame({
    'Comparison': list(hepg2_counts.keys()),
    'Rows': list(hepg2_counts.values()),
    'Unique CIDs': list(hepg2_unique_cids.values())
})
hepg2_df.loc[len(hepg2_df)] = ['TOTAL', sum(hepg2_counts.values()), sum(hepg2_unique_cids.values())]

# 3. Caspase vs Mito Comparisons
caspase_mito_df = pd.DataFrame({
    'Comparison': list(caspase_mito_counts.keys()),
    'Rows': list(caspase_mito_counts.values()),
    'Unique CIDs': list(caspase_mito_unique_cids.values())
})
caspase_mito_df.loc[len(caspase_mito_df)] = ['TOTAL', sum(caspase_mito_counts.values()), sum(caspase_mito_unique_cids.values())]

# 4. Mito vs Viability Comparisons
mito_viability_df = pd.DataFrame({
    'Comparison': list(mito_viability_counts.keys()),
    'Rows': list(mito_viability_counts.values()),
    'Unique CIDs': list(mito_viability_unique_cids.values())
})
mito_viability_df.loc[len(mito_viability_df)] = ['TOTAL', sum(mito_viability_counts.values()), sum(mito_viability_unique_cids.values())]

# 5. Caspase vs Mito vs Mito Viability HepG2 (8 Cases)
threeway_mito_viability_df = pd.DataFrame({
    'Case': list(threeway_mito_viability_counts.keys()),
    'Rows': list(threeway_mito_viability_counts.values()),
    'Unique CIDs': list(threeway_mito_viability_unique_cids.values())
})
threeway_mito_viability_df.loc[len(threeway_mito_viability_df)] = ['TOTAL', sum(threeway_mito_viability_counts.values()), sum(threeway_mito_viability_unique_cids.values())]

# 6. Caspase vs Mito vs Caspase Viability HepG2 (8 Cases)
threeway_caspase_viability_df = pd.DataFrame({
    'Case': list(threeway_caspase_viability_counts.keys()),
    'Rows': list(threeway_caspase_viability_counts.values()),
    'Unique CIDs': list(threeway_caspase_viability_unique_cids.values())
})
threeway_caspase_viability_df.loc[len(threeway_caspase_viability_df)] = ['TOTAL', sum(threeway_caspase_viability_counts.values()), sum(threeway_caspase_viability_unique_cids.values())]

# Export to Excel with multiple sheets
output_path = results_dir / 'comparison_results_summary.xlsx'
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    chok1_df.to_excel(writer, sheet_name='CHO-K1', index=False)
    hepg2_df.to_excel(writer, sheet_name='HepG2', index=False)
    caspase_mito_df.to_excel(writer, sheet_name='Caspase_vs_Mito', index=False)
    mito_viability_df.to_excel(writer, sheet_name='Mito_vs_Viability', index=False)
    threeway_mito_viability_df.to_excel(writer, sheet_name='Casp_Mito_MitoViab_8Cases', index=False)
    threeway_caspase_viability_df.to_excel(writer, sheet_name='Casp_Mito_CaspViab_8Cases', index=False)

print(f"Summary exported to: {output_path}")
print("\nSheets created:")
print("1. CHO-K1 - CHO-K1 Caspase vs Viability comparisons")
print("2. HepG2 - HepG2 Caspase vs Viability comparisons")
print("3. Caspase_vs_Mito - Caspase vs Mito HepG2 comparisons")
print("4. Mito_vs_Viability - Mito vs Mito Viability comparisons")
print("5. Casp_Mito_MitoViab_8Cases - Caspase vs Mito vs Mito Viability (8 cases)")
print("6. Casp_Mito_CaspViab_8Cases - Caspase vs Mito vs Caspase Viability (8 cases)")
```
