```python
import pandas as pd
import numpy as np
import glob
import os


print("Libraries loaded successfully!")
```

```python

base_path = "/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/"

files_to_load = [
    "AID_655_cleaned.csv",
    "AID_656_cleaned.csv",
    "AID_657_cleaned.csv",
    "AID_658_cleaned.csv",
    "AID_659_cleaned.csv",
    "AID_660_cleaned.csv",
    "AID_661_cleaned.csv",
    "AID_663_cleaned.csv",
    "AID_664_cleaned.csv",
    "AID_665_cleaned.csv",
    "AID_666_cleaned.csv",
    "AID_667_cleaned.csv",
    "AID_1346980_cleaned.csv",
    "combined_hepg2_casepase.csv"
]

cell_line = {}

for file_name in files_to_load:
    key = file_name.replace("_cleaned.csv", "")
    file_path = os.path.join(base_path, file_name)
    cell_line[key] = pd.read_csv(file_path,low_memory=False)

print(f"Loaded {len(cell_line)} datasets")

```

```python
for name, df in cell_line.items():
    print(name, df.shape)
```

```python
dfs = []

for name, df in cell_line.items():
    df_copy = df.copy()
    df_copy["source"] = name
    dfs.append(df_copy)

combined = pd.concat(dfs, ignore_index=True)


print(f"Loaded shape: {combined.shape}")
print(combined["source"].value_counts())
```

```python
# Save
combined.to_csv('combined_all_caspase_cell_to_one_casepase.csv', index=False)
```
