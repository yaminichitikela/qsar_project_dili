# Compare AID_720634 (Viability) vs AID_1346981 (Counter Screen)
Check if datatables can be merged by comparing columns and CID-outcome pairs

```python
import pandas as pd
```

```python
# Load both datasets
viability = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720634_cleaned.csv')
counter_screen = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_1346981_cleaned.csv')

print(f"Viability shape: {viability.shape}")
print(f"Counter Screen shape: {counter_screen.shape}")
```

## Step 1: Compare Column Names

```python
# Get column names
viability_cols = set(viability.columns)
counter_cols = set(counter_screen.columns)

# Common columns
common = viability_cols & counter_cols
print(f"Common columns ({len(common)}):")
print(sorted(common))

notcommon = viability_cols ^ counter_cols
print(f"Not common columns ({len(notcommon)}):")
print(sorted(notcommon))
```

```python
# Columns only in Viability
only_viability = viability_cols - counter_cols
print(f"Only in Viability ({len(only_viability)}):")
for col in sorted(only_viability)[:20]:  # Show first 20
    print(f"  {col}")
if len(only_viability) > 20:
    print(f"  ... and {len(only_viability) - 20} more")
```

```python
# Columns only in Counter Screen
only_counter = counter_cols - viability_cols
print(f"Only in Counter Screen ({len(only_counter)}):")
for col in sorted(only_counter)[:20]:  # Show first 20
    print(f"  {col}")
if len(only_counter) > 20:
    print(f"  ... and {len(only_counter) - 20} more")
```

## Step 2: Compare CID-Outcome Pairs

```python
# Merge on CID to compare outcomes
merged = pd.merge(
    viability[['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME','PUBCHEM_ACTIVITY_SCORE']].dropna(),
    counter_screen[['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME','PUBCHEM_ACTIVITY_SCORE']].dropna(),
    on='PUBCHEM_CID',
    suffixes=('_viability', '_counter')
)

# Find CIDs with different outcomes
different = merged[merged['PUBCHEM_ACTIVITY_OUTCOME_viability'] != merged['PUBCHEM_ACTIVITY_OUTCOME_counter']]
print(f"CIDs with different outcomes: {len(different)}")
print("\nOutcome distribution:")
print(different.groupby(['PUBCHEM_ACTIVITY_OUTCOME_viability', 'PUBCHEM_ACTIVITY_OUTCOME_counter']).size())
```

```python
# Show sample CIDs with different outcomes (including activity scores)
print("Sample CIDs with different outcomes:")
different[['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME_viability', 'PUBCHEM_ACTIVITY_SCORE_viability', 
           'PUBCHEM_ACTIVITY_OUTCOME_counter', 'PUBCHEM_ACTIVITY_SCORE_counter']].head(10)
```

## Step 3: Compare Column Values for a Specific CID

```python
# Pick one CID with different outcomes (change this to check any CID)
if len(different) > 0:
    example_cid = different['PUBCHEM_CID'].iloc[0]
else:
    example_cid = merged['PUBCHEM_CID'].iloc[0]

print(f"Checking CID: {example_cid}")
```

```python
# Get rows for this CID from both datasets
v_row = viability[viability['PUBCHEM_CID'] == example_cid].iloc[0]
c_row = counter_screen[counter_screen['PUBCHEM_CID'] == example_cid].iloc[0]

# Compare values for common columns (excluding Activity at concentration columns)
key_cols = ['PUBCHEM_CID', 'PUBCHEM_SID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE',
            'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']

comparison = pd.DataFrame({
    'Column': key_cols,
    'Viability': [v_row[col] if col in v_row else 'N/A' for col in key_cols],
    'Counter_Screen': [c_row[col] if col in c_row else 'N/A' for col in key_cols]
})
comparison
```

```python
# Show all common columns where values differ for this CID
diff_cols = []
for col in sorted(common):
    v_val = v_row[col] if col in v_row.index else None
    c_val = c_row[col] if col in c_row.index else None
    # Compare (handle NaN)
    if str(v_val) != str(c_val):
        diff_cols.append({'Column': col, 'Viability': v_val, 'Counter_Screen': c_val})

diff_df = pd.DataFrame(diff_cols)
print(f"Common columns with different values for CID {example_cid}: {len(diff_df)}")
diff_df.head(20)
```

## Step 4: Analysis - What's Driving Outcome Differences?

```python
# Compare Activity Scores between same and different outcome groups
same_outcomes = merged[merged['PUBCHEM_ACTIVITY_OUTCOME_viability'] == merged['PUBCHEM_ACTIVITY_OUTCOME_counter']]

print("=== Activity Score Comparison ===")
print(f"\nCIDs with SAME outcomes: {len(same_outcomes)}")
print(f"CIDs with DIFFERENT outcomes: {len(different)}")

print("\n--- Score differences for CIDs with DIFFERENT outcomes ---")
different['score_diff'] = abs(different['PUBCHEM_ACTIVITY_SCORE_viability'] - different['PUBCHEM_ACTIVITY_SCORE_counter'])
print(f"Mean score difference: {different['score_diff'].mean():.2f}")
print(f"Max score difference: {different['score_diff'].max():.2f}")
```

```python
# Merge full data to compare Phenotype and other key columns
key_columns = ['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE', 
               'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']

full_merged = pd.merge(
    viability[key_columns],
    counter_screen[key_columns],
    on='PUBCHEM_CID',
    suffixes=('_viability', '_counter')
)

# Compare Phenotypes
print("=== Phenotype Comparison (Replicate 1) ===")
phenotype_crosstab = pd.crosstab(
    full_merged['Phenotype-Replicate_1_viability'], 
    full_merged['Phenotype-Replicate_1_counter'],
    margins=True
)
print(phenotype_crosstab)
```

## Step 5: Summary & Merge Recommendations

```python
# Summary Statistics
total_cids = len(merged)
same = len(same_outcomes)
diff = len(different)

print("=" * 60)
print("SUMMARY: AID_720634 (Viability) vs AID_1346981 (Counter Screen)")
print("=" * 60)
print(f"\nBoth assays performed on: HepG2 cell line")
print(f"\n1. COLUMN COMPARISON:")
print(f"   - Common columns: {len(common)}")
print(f"   - Only in Viability: {len(only_viability)} (different concentration ranges)")
print(f"   - Only in Counter Screen: {len(only_counter)} (different concentration ranges)")

print(f"\n2. CID OVERLAP:")
print(f"   - Total CIDs in both datasets: {total_cids}")
print(f"   - CIDs with SAME outcome: {same} ({100*same/total_cids:.1f}%)")
print(f"   - CIDs with DIFFERENT outcome: {diff} ({100*diff/total_cids:.1f}%)")

print(f"\n3. KEY FACTORS DRIVING DIFFERENCES:")
print(f"   - Different assay endpoints (viability vs interference detection)")
print(f"   - Different concentration ranges tested")
print(f"   - Phenotype classifications differ (Cytotoxic vs Inactive, etc.)")

print(f"\n4. MERGE RECOMMENDATION:")
print(f"   - DO NOT simply combine outcomes - they measure different things")
print(f"   - KEEP both outcomes as separate columns with suffixes")
print(f"   - Use counter screen to FLAG potential false positives in viability")
```

```python
# Example: How to properly merge (keeping key columns from both)
# Select key columns you need from each dataset
viability_key = viability[['PUBCHEM_CID', 'PUBCHEM_SID', 'PUBCHEM_EXT_DATASOURCE_SMILES',
                           'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE',
                           'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']]

counter_key = counter_screen[['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE',
                               'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']]

# Merge with suffixes to keep both
combined = pd.merge(viability_key, counter_key, on='PUBCHEM_CID', 
                    suffixes=('_viability', '_counter'), how='outer')

print(f"Combined dataset shape: {combined.shape}")
print(f"\nColumns in combined dataset:")
print(combined.columns.tolist())
```

```python
# Preview the combined dataset
combined.head()
```

## Step 6: Export Two Sheets for Analysis

```python
# SHEET 1: All CIDs merged with common key columns
sheet1_all_merged = pd.merge(
    viability[['PUBCHEM_CID', 'PUBCHEM_SID', 'PUBCHEM_EXT_DATASOURCE_SMILES',
               'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE',
               'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']],
    counter_screen[['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE',
                    'Phenotype-Replicate_1', 'Potency-Replicate_1', 'Efficacy-Replicate_1']],
    on='PUBCHEM_CID',
    suffixes=('_viability', '_counter'),
    how='inner'
)

print(f"SHEET 1 - All Merged CIDs: {len(sheet1_all_merged)} rows")
sheet1_all_merged.head()
```

```python
# SHEET 2: Only CIDs with DIFFERENT outcomes (for analysis)
sheet2_different_outcomes = sheet1_all_merged[
    sheet1_all_merged['PUBCHEM_ACTIVITY_OUTCOME_viability'] != sheet1_all_merged['PUBCHEM_ACTIVITY_OUTCOME_counter']
].copy()

# Add score difference column for analysis
sheet2_different_outcomes['Score_Difference'] = abs(
    sheet2_different_outcomes['PUBCHEM_ACTIVITY_SCORE_viability'] - 
    sheet2_different_outcomes['PUBCHEM_ACTIVITY_SCORE_counter']
)

# Sort by score difference to see biggest discrepancies first
sheet2_different_outcomes = sheet2_different_outcomes.sort_values('Score_Difference', ascending=False)

print(f"SHEET 2 - Different Outcomes Only: {len(sheet2_different_outcomes)} rows")
sheet2_different_outcomes.head(10)
```

```python
# Export both sheets to Excel
output_path = '/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/viability_vs_counter_comparison.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    sheet1_all_merged.to_excel(writer, sheet_name='All_Merged_CIDs', index=False)
    sheet2_different_outcomes.to_excel(writer, sheet_name='Different_Outcomes', index=False)

print(f"Exported to: {output_path}")
print(f"\nSheet 1 (All_Merged_CIDs): {len(sheet1_all_merged)} rows")
print(f"Sheet 2 (Different_Outcomes): {len(sheet2_different_outcomes)} rows")
```
