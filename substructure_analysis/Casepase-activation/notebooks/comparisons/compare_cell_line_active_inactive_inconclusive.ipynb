# Summary Table - All Cell Lines

```python
# Create summary table
cell_lines = {
    'HepG2 Caspase': hepg2,
    'Jurkat': jurkat,
    'HUV-EC-C': huvec,
    'SH-SY5Y': shsy5y,
    'BJ': bj,
    'MRC-5': mrc5,
    'Mesangial': mesangial,
    'SK-N-SH': sknsh,
    'H-4-II-E': h4iie,
    'HEK293': hek293,
    'N2a': n2a,
    'NIH 3T3': nih3t3,
    'Renal Proximal Tubule': rpt,
    'CHO-K1 Caspase': chok1,
    'Viability CHO-K1': viability_chok1,
    'Viability HepG2 Caspase': viability_hepg2_caspase,
    'Viability HepG2 Mito': viability_hepg2_mito,
    'Mito Toxicity HepG2': mito_hepg2
}

summary_data = []
for name, df in cell_lines.items():
    counts = df['PUBCHEM_ACTIVITY_OUTCOME'].value_counts()
    active = counts.get('Active', 0)
    inactive = counts.get('Inactive', 0)
    inconclusive = counts.get('Inconclusive', 0)
    total = df['PUBCHEM_CID'].nunique()
    summary_data.append({
        'Dataset': name,
        'Total Compounds': total,
        'Active': active,
        'Inactive': inactive,
        'Inconclusive': inconclusive,
        'Active %': round(active / len(df) * 100, 2) if len(df) > 0 else 0,
        'Inactive %': round(inactive / len(df) * 100, 2) if len(df) > 0 else 0
    })

summary_df = pd.DataFrame(summary_data)
print("=" * 100)
print("SUMMARY: Active/Inactive/Inconclusive Counts for All Datasets")
print("=" * 100)
print(summary_df.to_string(index=False))
print("=" * 100)
```
