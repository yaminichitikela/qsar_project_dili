# Compare Caspase (HepG2), Mitochondrial Toxicity (HepG2), and mito Viability (HepG2)

This notebook compares three datasets:

- **AID 720634**: Viability HepG2 (Viability)
- **Combined HepG2 Caspase**: Caspase-3/7 activation HepG2 (Caspase)

**Eight Cases Analysis (All Combinations):**
1. Case 1: Caspase Active, Mito Active, Viability Active
2. Case 2: Caspase Active, Mito Active, Viability Inactive
3. Case 3: Caspase Active, Mito Inactive, Viability Active
4. Case 4: Caspase Active, Mito Inactive, Viability Inactive

6. Case 6: Caspase Inactive, Mito Active, Viability Inactive

```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

## Step 1: Load AID 720635 (Mito Toxicity) and Combined HepG2 Caspase Data

```python
# Load AID 720635 - Mitochondrial Toxicity HepG2
aid_720635_mito = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720635_cleaned.csv', 
                       low_memory=False)

# Load Combined HepG2 Caspase data
caspase_hepg2 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/combined_hepg2_casepase.csv', 
                       low_memory=False)

mito_viability = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_720634_cleaned.csv', 
                       low_memory=False)

print("AID 720635 - Mitochondrial Toxicity HepG2")
print(f"Loaded shape: {aid_720635_mito.shape}")
print(f"Unique CIDs: {aid_720635_mito['PUBCHEM_CID'].nunique()}")

print("\nCombined HepG2 Caspase")
print(f"Loaded shape: {caspase_hepg2.shape}")
print(f"Unique CIDs: {caspase_hepg2['PUBCHEM_CID'].nunique()}")

print("\nCombined HepG2 Caspase")
print(f"Loaded shape: {mito_viability.shape}")
print(f"Unique CIDs: {mito_viability['PUBCHEM_CID'].nunique()}")
```

```python
# Sort by CID and activity score (highest first)
aid_720635_mito_sorted = aid_720635_mito.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)
caspase_hepg2_sorted = caspase_hepg2.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)
mito_viability_sorted = mito_viability.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

# Create filtered datasets for Active and Inactive
filtered_caspase_active = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_caspase_inactive = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

filtered_mito_active = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_mito_inactive = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

filtered_viability_active = mito_viability_sorted[mito_viability_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Active']
filtered_viability_inactive = mito_viability_sorted[mito_viability_sorted['PUBCHEM_ACTIVITY_OUTCOME'] == 'Inactive']

# Get CID sets for intersection
caspase_active_cids = set(filtered_caspase_active['PUBCHEM_CID'])
caspase_inactive_cids = set(filtered_caspase_inactive['PUBCHEM_CID'])
mito_active_cids = set(filtered_mito_active['PUBCHEM_CID'])
mito_inactive_cids = set(filtered_mito_inactive['PUBCHEM_CID'])
viability_active_cids = set(filtered_viability_active['PUBCHEM_CID'])
viability_inactive_cids = set(filtered_viability_inactive['PUBCHEM_CID'])

print("Datasets prepared for 8-case analysis")
```

```python
# CASE 1: Caspase Active, Mito Active, Viability Active
case1_common_cids = caspase_active_cids.intersection(mito_active_cids).intersection(viability_active_cids)
print(f"Case 1 - Caspase Active, Mito Active, Viability Active: {len(case1_common_cids)} compounds")

# Get rows for Case 1 compounds from all three datasets
case1_caspase = caspase_hepg2_sorted[caspase_hepg2_sorted['PUBCHEM_CID'].isin(case1_common_cids)]
case1_mito = aid_720635_mito_sorted[aid_720635_mito_sorted['PUBCHEM_CID'].isin(case1_common_cids)]
case1_viability = mito_viability_sorted[mito_viability_sorted['PUBCHEM_CID'].isin(case1_common_cids)]

# Combine and save Case 1
case1_merged = pd.concat([case1_caspase, case1_mito, case1_viability], ignore_index=True)
case1_merged.to_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/results/casepase_mito_comparisons/case1_caspase_active_mito_active_viability_active.csv', index=False)
print("Case 1 saved.")
```
