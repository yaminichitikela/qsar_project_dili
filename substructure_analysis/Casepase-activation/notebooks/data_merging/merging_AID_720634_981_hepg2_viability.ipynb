```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

```python
casepase_aid_720634 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_1346978_cleaned.csv', 
                       low_memory=False)
casepase_aid_981 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_664_cleaned.csv', 
                       low_memory=False)

print(f"Loaded shape: {casepase_aid_720634.shape}")
print(f"Loaded shape: {casepase_aid_981.shape}")


casepase_aid_981.head().sort_values(by='PUBCHEM_CID', ascending=True)
```

```python

combined = pd.concat([casepase_aid_720634, casepase_aid_981], ignore_index=True)

print(f"Loaded shape: {combined.shape}")

print(combined.head().sort_values(by='PUBCHEM_CID', ascending=True))
```

```python
# Remove duplicates within mito: keep highest score per CID+outcome
hepg2_sorted = combined.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

hepg2_viability_clean = hepg2_sorted.drop_duplicates(
    subset=['PUBCHEM_CID'],
    keep='first'
).copy()

print(f"After dedup: {combined.shape} -> {hepg2_viability_clean.shape}")
print(f"Removed {len(combined) - len(hepg2_viability_clean)} duplicate rows")

print(hepg2_viability_clean.head())


# Verify: Check CID 6
print("Verification - CID 4:")
print("\nBefore dedup (combined):")
display(combined[combined['PUBCHEM_CID'] == 403][['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE']])

print("\nAfter dedup (merged):")
display(hepg2_viability_clean[hepg2_viability_clean['PUBCHEM_CID'] == 403][['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE']])
```
