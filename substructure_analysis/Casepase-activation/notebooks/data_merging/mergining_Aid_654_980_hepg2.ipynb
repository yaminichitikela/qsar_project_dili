```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

```python
# Load AID 720634 - skip first 5 metadata rows
casepase_aid_980 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_1346978_cleaned.csv', 
                       low_memory=False)
casepase_aid_654 = pd.read_csv('/Users/yaminichitikela/Documents/casepase/Casepase-activation/cleaned_datasets/AID_654_cleaned.csv', 
                       low_memory=False)

print("AID 1346981 - casepase HepG2")
print(f"Loaded shape: {casepase_aid_980.shape}")
print(f"Loaded shape: {casepase_aid_654.shape}")

```

```python

combined = pd.concat([casepase_aid_980, casepase_aid_654], ignore_index=True)

print(f"Loaded shape: {combined.shape}")

print(combined.head())

```

```python
combined.to_csv('concat_hepg2_cleaned.csv', index=False)
```

```python
# Remove duplicates within mito: keep highest score per CID+outcome
mito_sorted = combined.sort_values(
    by=['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_SCORE'],
    ascending=[True, False]
)

mito_clean = mito_sorted.drop_duplicates(
    subset=['PUBCHEM_CID'],
    keep='first'
).copy()

print(f"After dedup: {combined.shape} -> {mito_clean.shape}")
print(f"Removed {len(combined) - len(mito_clean)} duplicate rows")

print(mito_clean.head())


# Verify: Check CID 6
print("Verification - CID 4:")
print("\nBefore dedup (combined):")
display(combined[combined['PUBCHEM_CID'] == 403][['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE']])

print("\nAfter dedup (merged):")
display(mito_clean[mito_clean['PUBCHEM_CID'] == 403][['PUBCHEM_CID', 'PUBCHEM_ACTIVITY_OUTCOME', 'PUBCHEM_ACTIVITY_SCORE']])
```

```python
# Save
mito_clean.to_csv('combined_hepg2_casepase.csv', index=False)
```
