```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully!")
```

```python
# Load AID_654 (NCATS HepG2) -- skip the 3 metadata rows (RESULT_TYPE, RESULT_DESCR, RESULT_UNIT)
aid_654 = pd.read_csv(
    '/Users/yamini/Desktop/projects/qsar casepase/qsar_prediction/Rebuttal/ncats_13_cell_types/AID_654_datatable.csv',
    skiprows=[1, 2, 3],
    low_memory=False)

# Load AID_1347034 -- skip the 3 metadata rows (RESULT_TYPE, RESULT_DESCR, RESULT_UNIT)
aid_1347034 = pd.read_csv(
    '/Users/yamini/Desktop/projects/qsar casepase/qsar_prediction/Rebuttal/summary datasets/AID_1347034_datatable.csv',
    skiprows=[1, 2, 3],
    low_memory=False)

print("AID_654 (NCATS HepG2)")
print(f"Loaded shape: {aid_654.shape}")
print()
print("AID_1347034")
print(f"Loaded shape: {aid_1347034.shape}")
```

```python
# Merge only -- stack the two datasets together. No deduplication here;
# that's handled separately by the cleaning notebook (average score, classify
# outcome from the averaged score).
combined = pd.concat([aid_654, aid_1347034], ignore_index=True)

print(f"Combined shape: {combined.shape}")
print(f"Total rows: {len(aid_654)} + {len(aid_1347034)} = {len(aid_654) + len(aid_1347034)}")
print(combined.head())
```
